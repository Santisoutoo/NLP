### 1. Case folding

```python
with open('texts/viaje_centro_tierra.txt', 'r') as chapter:
    text = chapter.read()
    text_lower = text.lower()
```

## Word normalization

### 3. Tokenization

```python
from nltk.tokenize import word_tokenize
```

```python
CHAPTER_TOKENS = word_tokenize(text_lower)
print(CHAPTER_TOKENS)
```

### 4. Stemming

```python
from nltk.stem import SnowballStemmer
```

```python
stemmer = SnowballStemmer('spanish')

stems = []
for word in CHAPTER_TOKENS:
    stems.append(stemmer.stem(word))
print(stems)
```
