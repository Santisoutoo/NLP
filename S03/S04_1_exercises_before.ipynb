### 1 Exercise: Split the file cantares_gallegos.txt into words. 

```python
with open('texts/cantares_gallegos.txt', 'r') as file:
    words = file.read().split()
    print(words)
    
```

### 2 Exercise: Count the number of ocurrences of each word in the file cantares_gallegos.txt.

```python
def word_count(sentence: list):
    word_counts = {}
    for word in sentence:
        if word in word_counts:
            word_counts[word] += 1
        else:
            word_counts[word] = 1
    return word_counts
```

```python
with open('texts/cantares_gallegos.txt', 'r') as file:
    counting = word_count(file)
    print(counting)
```
