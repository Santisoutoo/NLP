###  3 - Convert to lowercase except the proper nouns that must keep the original case. For the multiword proper names convert them to an unique word joining the two word with underscoere (Juan Fernández -> Juan_Fernández). (1 point)


```python
proper_names = {
    'Juan Fernández': 'Juan_fernandez'
}
text_changed = []

for word in TEXT:
    if word in proper_names:
        text_changed.append(proper_names)
    else: text_changed.append(word.lower())

print(text_changed)
```

###  4 - Tokenize the text (use the tool you prefer). (1 point)


```python
tokens = nltk.word_tokenize(TEXT)

print("Tokenized text:", tokens)
```

###  5 - Remove the stopwords (use the tool you prefer). (1 point)


```python
from nltk.corpus import stopwords

stop_words = stopwords.words(LANGUAGE) 

filtered_words = []
for word in tokens:
    if word.lower() not in stop_words:
        filtered_words.append(word)

print("Text with no stopwords:", filtered_words)
```

###  6 - Create bigrams with pure python. (1 point)


```python
bigrams = []

# we iterate thourt the list indexes with a step = 1
# and we make pairs. the second word of the bigram will be allways the first
# of following one
for i in range(len(filtered_words)-1):
    bigrams.append((filtered_words[i], filtered_words[i+1]))

print(bigrams)
```
