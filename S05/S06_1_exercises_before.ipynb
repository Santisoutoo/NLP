Create a corpus with phrases and annotate it with positive or negative. Build a very simple sentiment analysis with regex and use corpus to calculate sentiment (positive or negative). Count how many True Positives (TP), False Positives (FP), True Negatives (TN), and False Negatives (FN) you have. Tweet the corpus or the regexp to have non-zero variables. Make the calculations of Precission, Recall, and F1 following the class materials.   

```python
import re

# Create a corpus with annotated sentences (1 for positive, 0 for negative)
corpus = [
    ("just plain boring", 0),
    ("entirely predictable and lacks energy", 0),
    ("no surprises and very few laughs", 0),
    ("very powerful", 1),
    ("the most fun film of the summer", 1)
]
```

```python
positive_patterns = [
    r'powerful|fun|great|good|best|summer'
]
negative_patterns = [
    r'boring|predictable|lacks|no surprises|few laughs'
]
```

```python
# Evaluate predictions
TP = FP = TN = FN = 0

for text, actual in corpus:
    # Simple prediction: if any positive word and no negative word -> positive
    has_positive = False
    has_negative = False
    
    # Check positive patterns
    for pattern in positive_patterns:
        for word in pattern.split('|'):
            if word in text.lower():
                has_positive = True
                break
        if has_positive:
            break
            
    # Check negative patterns
    for pattern in negative_patterns:
        for word in pattern.split('|'):
            if word in text.lower():
                has_negative = True
                break
        if has_negative:
            break
    
    predicted = 1 if has_positive and not has_negative else 0
    
    if predicted == 1 and actual == 1: TP += 1
    elif predicted == 1 and actual == 0: FP += 1
    elif predicted == 0 and actual == 0: TN += 1
    elif predicted == 0 and actual == 1: FN += 1

```

```python
# Calculate metrics
precision = TP / (TP + FP) if (TP + FP) > 0 else 0
recall = TP / (TP + FN) if (TP + FN) > 0 else 0
f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

print(f"True Positives: {TP}")
print(f"False Positives: {FP}")
print(f"True Negatives: {TN}")
print(f"False Negatives: {FN}")
print(f"\nPrecision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"F1 Score: {f1:.2f}")
```
