6. Write a function that takes two strings and checks if they are anagrams.

```python
def is_anagram(word1: str, word2: str):
    if len(word1) != len(word2):
        return False
    else:
        if sorted(word1) == sorted(word2):
            return True
        else:
            return False

print(is_anagram('hola', 'halo')) 
```

7. Implement a function that removes all vowels from a given string.

```python
def remove_vogal(word: str):
    vogals = ('a', 'e', 'i', 'o', 'u')
    lista = list(word)

    i = 0
    while i < len(lista):
        if lista[i] in vogals:
            lista.pop(i)  
        else:
            i = i + 1
    return lista

print(remove_vogal(INPUT))  
```

8. Write a function that checks if a string is a palindrome (case-insensitive).

```python
def is_palindrome(word: str):
    word.lower()
    if word == word[::-1]:
        return True
    else:
        return False

print(is_palindrome('AnA'))
```

3. Implement a function to count the occurrences of each word in a given sentence.

```python
def word_count(sentence):
    words = sentence.split()
    word_counts = {}
    for word in words:
        if word in word_counts:
            word_counts[word] += 1
        else:
            word_counts[word] = 1
    return word_counts

sentence = "test test caracola caracola hola"
print(word_count(sentence))
```

11. Implement a function to find the intersection of two sets.

```python
def intersection(set1, set2):
    common_elements = set()
    for item in set1:
        if item in set2:
            common_elements.add(item)
    return common_elements

print(intersection(
    {1, 2, 3, 4},
    {3, 4, 5, 6}
))
```

9. Write a program to merge two DataFrames on a common column.

```python
df1 = pd.DataFrame(
    {
        'id_number': [1, 2, 3], 
        'Name': ['Alice', 'Bob', 'Charlie']
    }
)
df2 = pd.DataFrame(
    {
        'id_number': [1, 2, 4], 
        'poitns': [85, 90, 95]
    }
)

# It is based on the same idea used in SQL.
for type in ['inner', 'left', 'right', 'outer']:
    merged_df = pd.merge(
        df1, df2, 
        on = 'id_number', 
        how = type
    )
    display(merged_df)
```
