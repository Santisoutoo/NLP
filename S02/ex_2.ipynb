In regular expressions, ?: is used for non-capturing groups. It allows you to group parts of a pattern without storing the match. This is useful when you want to apply quantifiers (like ?, *, {n,m}) but don't need to capture the matched content.

    (?:...): Groups the pattern but doesn't save the match.
    ?: Makes the group optional (can appear 0 or 1 times).

Example:

(?:\.[a-zA-Z0-9]{1,3})?

This pattern matches an optional file extension (like .txt, .py, etc.) without capturing it. It looks for a dot followed by 1 to 3 alphanumeric characters, but it doesn't keep the match for further use.
How findall Works:

The re.findall() function in Python is used to find all non-overlapping matches of a pattern in a string. It returns a list of the matches found. If the pattern contains capturing groups, only the parts of the string that are matched by those groups are returned.
How It Works:

    findall with capturing groups: Returns only the captured groups.
    findall with non-capturing groups (?:): Will still match the entire pattern, but the parts of the pattern enclosed in non-capturing groups will not be included in the results.

Example:

```Python
extension = '.txt .py .html.d'

# Pattern with non-capturing group for optional second extension
r = re.findall(r'\.[a-zA-Z0-9]{1,5}(?:\.[a-zA-Z0-9]{1,3})?', extension)

print(r)
```
Output:

['.txt', '.py', '.html', '.d']

Here, findall returns the full match, which includes the initial dot and alphanumeric characters (such as .txt, .py), but the non-capturing group for the optional second extension (e.g., .d) is not included in the result. This demonstrates how ?: helps in grouping patterns without affecting the output of findall.

### Exercise 17: Validate a URL
Write a regex to validate a URL (e.g., https://www.example.com).


````python
url = 'https://www.google.es http://www.amazon.com https://www.github.org http://www.microsoft.com'
r = re.findall(r'http[s]?:\/\/www.[a-z]+.(?:com|es|org)', url)
print(r)
````
