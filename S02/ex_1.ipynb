## Exercise 1: Make Plurals

Write a program to generate plural forms of English nouns from a given list of singular nouns.

### Simplification Rules:
1. **Add 's'** to most words  
    *Example:* `cat` -> `cats`
2. If the word ends with **'y'** and is preceded by a consonant, replace **'y'** with **'ies'**  
    *Example:* `baby` -> `babies`
3. If the word ends with **'f'** or **'fe'**, replace it with **'ves'**  
    *Example:* `wolf` -> `wolves`
4. **Add 'es'** to words ending in **'s'**, **'x'**, **'z'**, **'ch'**, or **'sh'**  
    *Example:* `box` -> `boxes`
5. For **irregular nouns**, handle exceptions  
    *Examples:* `child` -> `children`, `man` -> `men`

```python
NOUNS = ["cat", "baby", "wolf", "box", "child", "man", "mouse", "knife", "bush"]
```

```python
def words_to_plural(nouns: list):
    plurals = []
    consonantes = 'bcdfghjklmnpqrstvwxyz'
    
    for word in nouns:
        length = len(word)

        if word[length-1] == 'y' and word[length-2] in consonantes:
            plurals.append(word[:length - 1] + 'ies')

        elif word[length-1] == 'f': 
            plurals.append(word[:length - 1] + 'ves')

        elif length > 1 and word[length-2:length] == 'fe': 
            plurals.append(word[:length - 2] + 'ves')

        elif (
            word[length-1] == 's' or 
            word[length-1] == 'x' or 
            word[length-1] == 'z' or 
            word[length-2:length] == 'ch' or 
            word[length-2:length] == 'sh'
        ):

            plurals.append(word + 'es')

        elif word == 'child':
            plurals.append('children')
        elif word == 'man':
            plurals.append('men')

        else: 
            plurals.append(word + 's')

    return plurals
```

```python
print(words_to_plural(NOUNS))
```

## Exercise 2: Extract Email Addresses

Write a Python program to extract all email addresses from the following string.

```python
TEXT = "Please contact us at support@example.com, sales@example.org, or admin123@example.net for assistance."
```

```python
import re

def get_email(text):
    # Expresión regular para encontrar direcciones de correo electrónico
    email_pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b"
    
    # Buscar todas las direcciones de correo en el texto
    emails = re.findall(email_pattern, text)
    
    return emails
```

```python
print("Emails encontrados:", get_email(TEXT))
```
