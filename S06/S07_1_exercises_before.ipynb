```python
import nltk
from nltk import ngrams
from collections import Counter

def analyze_ngrams(text, n):
    tokens = nltk.word_tokenize(text)
    n_grams = list(ngrams(tokens, n))
    freq_dist = Counter(n_grams)
    return freq_dist.most_common(10)
```

```python
# Test with different texts and n-gram sizes
def test_ngrams():
    # Make sure we have the required NLTK data
    nltk.download('punkt')
    
    # Test case 1: Simple sentence with bigrams
    text1 = "The quick brown fox jumps over the lazy dog."
    print("Bigrams from simple sentence:")
    print(analyze_ngrams(text1, 2))
    
    # Test case 2: Repeated phrases with trigrams
    text2 = "I love this movie. I love the actors. This movie is great."
    print("\nTrigrams from repeated phrases:")
    print(analyze_ngrams(text2, 3))
    
    # Test case 3: Movie review with 4-grams
    from nltk.corpus import movie_reviews
    review = movie_reviews.raw(fileids=movie_reviews.fileids()[0])
    print("\n4-grams from movie review:")
    print(analyze_ngrams(review, 4))

# Run the tests
test_ngrams()
```

```python
from nltk.lm.preprocessing import padded_everygram_pipeline
from nltk.lm import MLE
import nltk

def train_language_model(text, n=2):
    tokenized_text = [nltk.word_tokenize(sent) for sent in nltk.sent_tokenize(text)]
    train_data, vocab = padded_everygram_pipeline(n, tokenized_text)
    
    model = MLE(n)
    model.fit(train_data, vocab)
    return model
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer
from nltk import ne_chunk, pos_tag, word_tokenize
from nltk.tree import Tree
```

```python
def analyze_entities(text):
    """
    Perform Named Entity Recognition using NLTK
    """
    # Tokenize and tag the words
    tokens = word_tokenize(text)
    tagged = pos_tag(tokens)
    
    # Extract named entities
    entities = []
    chunks = ne_chunk(tagged)
    
    for chunk in chunks:
        if isinstance(chunk, Tree):
            entity = ' '.join(c[0] for c in chunk.leaves())
            entity_type = chunk.label()
            entities.append((entity, entity_type))
    
    return entities
```

```python
def analyze_sentiment(text):
    """
    Perform sentiment analysis using NLTK's VADER
    """
    sia = SentimentIntensityAnalyzer()
    return sia.polarity_scores(text)
```

```python
# Sample text for testing
text = """
Natural Language Processing (NLP) is a fascinating field of artificial intelligence.
It helps computers understand and process human language effectively.
Companies like Google and Amazon use NLP extensively in their products.
"""

# Test different functionalities
print("1. Bigrams Analysis:")
print(analyze_ngrams(text, 2))

print("\n2. Named Entities:")
print(analyze_entities(text))

print("\n3. Sentiment Analysis:")
print(analyze_sentiment(text))
```

```python
from nltk.corpus import wordnet

def explore_word_relations(word):
    syns = wordnet.synsets(word)
    if not syns:
        return f"No synsets found for '{word}'"
    
    result = []
    for syn in syns:
        result.append(f"\nSynset: {syn.name()}")
        result.append(f"Definition: {syn.definition()}")
        result.append("Synonyms: " + ", ".join([lemma.name() for lemma in syn.lemmas()]))
        result.append("Hypernyms: " + ", ".join([hyper.name() for hyper in syn.hypernyms()]))
        result.append("Hyponyms: " + ", ".join([hypo.name() for hypo in syn.hyponyms()]))
    
    return "\n".join(result)
```

```python
import nltk
# Download required WordNet data
nltk.download('wordnet')

# Test with different words
test_words = ['computer', 'happy', 'run', 'python']

for word in test_words:
    print(f"\n{'='*50}")
    print(f"Exploring word: {word}")
    print(f"{'='*50}")
    print(explore_word_relations(word))
```
