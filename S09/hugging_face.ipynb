```python
import os
import IPython.display as ipd
import soundfile as sf

# Define the directory to save audio files
audio_directory = "./audio"

# Create the directory if it doesn't exist
if not os.path.exists(audio_directory):
    os.makedirs(audio_directory)

# Loop through the first 3 audio samples
for i in range(3):
    print(f"Audio sample {i+1}:")
    sample = ds['train']['audio'][i]
    
    # Define the output file path
    file_path = os.path.join(audio_directory, f"sample_{i+1}.wav")
    
    # Save the audio sample as a WAV file
    sf.write(file_path, sample["array"], sample["sampling_rate"])
    
    # Display the audio player using the saved file
    ipd.display(ipd.Audio(file_path))
```
