```python
import pyttsx3
```

```python
engine = pyttsx3.init()
engine.setProperty("rate", 150)
engine.setProperty("volume", 0.9)
```

```python
from pydub import AudioSegment

# Save speech output to a WAV file
engine.save_to_file("ROGER.", "audio/output.wav")
engine.runAndWait()

# Convert the WAV file to MP3 using pydub (ensure pydub and ffmpeg are installed)

sound = AudioSegment.from_wav("audio/output.wav")
sound.export("output.mp3", format="mp3")
```
