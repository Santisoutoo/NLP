# Speach to text API

```python
import whisper
from datasets import load_dataset
```

```python
# 1. Load a Whisper model. Options: tiny, base, small, medium, large
model = whisper.load_model("tiny")
```

```python
ds = load_dataset("jacktol/atc-dataset")
print(ds)
```

```python
# 2. Path to your audio file (e.g., WAV, MP3)
audio_path = ""

# 3. Transcribe the audio
#    Specify language code if you know it (e.g., 'es' for Spanish)
result = model.transcribe(audio_path, language="es")
```

```python
# 4. Extract and print the transcribed text
transcript = result.get("text", "")
print("Transcription:", transcript)
```
