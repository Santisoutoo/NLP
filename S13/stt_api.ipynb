# Speech-to-Text with Whisper and ATC Dataset
This notebook demonstrates how to use OpenAI's Whisper model to transcribe the first few samples from the ATC dataset.

## Prerequisites

- Install **FFmpeg** on your system so Whisper can read/write audio files.
- Install Python packages below via pip.

```python
import whisper
from datasets import load_dataset
import pandas as pd
from IPython.display import display
```

```python
ds = load_dataset('jacktol/atc-dataset')
train_ds = ds['train']
print(f"Loaded 'train' split with {len(train_ds)} samples.")
```

```python

model = whisper.load_model('tiny')
print(model)
```

```python
N = 5
records = []
for i, sample in enumerate(train_ds[0:N], start=1):
    audio_path = sample['audio']['path']
    res = model.transcribe(audio_path, language='en')
    records.append({
        'sample': i,
        'reference': sample['text'],
        'whisper': res['text'].strip()
    })

df = pd.DataFrame(records)
display(df)
```

## Save results to CSV

```python
df.to_csv('whisper_atc_results.csv', index=False)
print('Results saved to whisper_atc_results.csv')
```
