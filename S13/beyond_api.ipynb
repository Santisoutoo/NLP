# Language recognition

```python
import langid
```

```python
def identify_language(text: str):
    lang, score = langid.classify(text)
    print(f"Detected language: {lang} (confidence {score:.2f})")
    return lang, score
```

```python

eg01 = "Bonjour tout le monde"

print("Example 1:")
identify_language(eg01)

print("\nExample 2:")    
# 3) Identify from user input
user_text = input("Enter some text to detect its language: ").strip()
if user_text:
    identify_language(user_text)
else:
    print("No text entered, exiting.")
```
